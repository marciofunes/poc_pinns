```python
import deepxde as dde
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import json
import os

os.environ['DDE_BACKEND'] = 'paddle' 

# --- 1. Processamento Eficiente do Dataset de 63MB --- #
json_file_path = 'dataset-31359-main.json'

def load_and_process_data(path):
    print(f"Carregando dataset de {path}...")
    with open(path, 'r') as f:
        data = json.load(f)
    
    df = pd.DataFrame(data)
    
    # Conversão de tipos e limpeza
    for col in ["PIT-5412001", "PIT-5412014", "FIT-5412014", "TIT-5412014"]:
        df[col] = pd.to_numeric(df[col], errors='coerce')
    
    # Tratamento de outliers/erros (ex: pressões negativas em PIT-5412001)
    # No seu dado, PIT-5412001 está em torno de -0.25. 
    # Vamos assumir que isso é uma pressão manométrica próxima de zero ou vácuo.
    # Para a física do duto, trabalharemos com a diferença de pressão.
    p_in_mean = df["PIT-5412001"].mean()
    p_out_mean = df["PIT-5412014"].mean()
    
    print(f"Estatísticas do Dataset:")
    print(f" - Pressão Entrada (Média): {p_in_mean:.4f}")
    print(f" - Pressão Saída (Média): {p_out_mean:.4f}")
    print(f" - Total de amostras processadas: {len(df)}")
    
    return p_in_mean, p_out_mean

p_in, p_out = load_and_process_data(json_file_path)

# --- 2. Definição do Modelo Físico (PINN) --- #
L = 10.0 # Comprimento do duto em metros
geom = dde.geometry.Interval(0, L)

# Equação de Darcy-Weisbach simplificada para perfil de pressão:
# dP/dx = -f * (rho/2) * (v²/D) -> d²P/dx² = 0 (para fluxo constante)
def pde(x, y):
    dy_xx = dde.grad.hessian(y, x)
    return dy_xx

# Condições de Contorno baseadas nos seus dados reais
bc_in = dde.icbc.DirichletBC(geom, lambda x: p_in, lambda x, on_boundary: np.isclose(x[0], 0))
bc_out = dde.icbc.DirichletBC(geom, lambda x: p_out, lambda x, on_boundary: np.isclose(x[0], L))

# Dados de observação (usando a média para guiar a rede)
observe_x = np.array([[0], [L]])
observe_y = np.array([[p_in], [p_out]])
observe_bc = dde.icbc.PointSetBC(observe_x, observe_y, component=0)

data_pinn = dde.data.PDE(geom, pde, [bc_in, bc_out], num_domain=2000, num_boundary=2)

# --- 3. Arquitetura e Treinamento --- #
net = dde.nn.FNN([1] + [40] * 3 + [1], "tanh", "Glorot uniform")

# Modelo PINN
model_pinn = dde.Model(data_pinn, net)
model_pinn.compile("adam", lr=1e-3)
model_pinn.train(iterations=15000)

# Modelo IA Convencional (Black-box)
# Treinada apenas nos pontos de entrada/saída para mostrar a falha na interpolação sem física
data_nn = dde.data.DataSet(X_train=observe_x, y_train=observe_y, X_test=observe_x, y_test=observe_y)
model_nn = dde.Model(data_nn, dde.nn.FNN([1] + [40] * 3 + [1], "tanh", "Glorot uniform"))
model_nn.compile("adam", lr=1e-3)
model_nn.train(iterations=15000)

# --- 4. Comparação e Visualização --- #
x_plot = np.linspace(0, L, 100).reshape(-1, 1)
y_pinn = model_pinn.predict(x_plot)
y_nn = model_nn.predict(x_plot)

plt.figure(figsize=(10, 6))
plt.plot(x_plot, y_pinn, label="PINN (Física + Dados)", color='blue', lw=2)
plt.plot(x_plot, y_nn, label="IA Convencional (Apenas Dados)", color='red', linestyle='--')
plt.scatter([0, L], [p_in, p_out], color='green', label="Sensores Reais (Média)", zorder=5)
plt.title("Diferencial PINN: Inferência de Perfil de Pressão com Dados Esparsos")
plt.xlabel("Posição no Duto (m)")
plt.ylabel("Pressão (PIT)")
plt.legend()
plt.grid(alpha=0.3)
plt.savefig("resultado_poc_63mb.png")
plt.show()
```
